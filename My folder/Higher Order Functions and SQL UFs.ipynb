```sql
Create or replace function get_url(email string)
returns string

return concat("https://wwww.",split(email, "@")[1])
```

```sql
select customer_id, get_url(email) as domain
from customers
```

```sql
create or replace function site_type(email string)
returns string
return Case 
            When email like '%.com%' then 'Comercial'
            when email like '%.org%' then 'Oreganization'
            when email like '%.edu%' then 'Educational'
        else concat("Unknown extension", split(email, '@')[1])
end;
```

```sql
select customer_id, site_type(email) as site 
from customers
```
